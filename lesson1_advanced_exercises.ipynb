## Exercise 1: Variable swap and trace

- Create `a = 15` and `b = 40`.
- Swap them **without** using a third variable.
- Print before/after values in a clear format.
- Print the type of each variable after swapping.

```python
# 1. Initialize variables
a = 15
b = 40

print(f"Before Swap: a = {a}, b = {b}")

# 2. Swap without using a third variable (arithmetic method)
a = a + b  # a = 55
b = a - b  # b = 15
a = a - b  # a = 40

# 3. Print values after swapping
print(f"After Swap:  a = {a}, b = {b}")

# 4. Print types
print(f"Type of a: {type(a)}")
print(f"Type of b: {type(b)}")
```

## Exercise 2: Dynamic typing checkpoint

- Create one variable `v` and assign it sequentially: string, int, float, bool, list.
- After each assignment, print the value and its type.
- Explain in a comment what this shows about Python variables.

```python
# Reassign 'v' sequentially across different types
v = "Hello, Python!"
print(f"Value: {v:<20} | Type: {type(v)}")

v = 100
print(f"Value: {v:<20} | Type: {type(v)}")

v = 99.99
print(f"Value: {v:<20} | Type: {type(v)}")

v = True
print(f"Value: {v:<20} | Type: {type(v)}")

v = [1, 2, 3]
print(f"Value: {v!s:<20} | Type: {type(v)}")
```

## Exercise 3: Parallel assignment and unpacking

- Assign `x, y, z` in one line with three different data types.
- Assign the same value to `a`, `b`, and `c` in one line.
- Create a list of 3 values and unpack it into three variables.
- Print all variables and the type of `x`, `y`, and `z`.

```python
# 1. Assign 3 variables in one line with 3 different data types
x, y, z = 42, 3.14, "Python"

# 2. Assign the same value to three variables in one line
a = b = c = 100

# 3. Unpack a list of 3 values into three variables
data_list = ["apple", "banana", "cherry"]
item1, item2, item3 = data_list

# 4. Print all variables and types for x, y, z
print("--- Variables ---")
print(f"x = {x}, y = {y}, z = '{z}'")
print(f"a = {a}, b = {b}, c = {c}")
print(f"items = {item1}, {item2}, {item3}")

print("\n--- Types of x, y, z ---")
print(f"Type of x: {type(x)}")
print(f"Type of y: {type(y)}")
print(f"Type of z: {type(z)}")
```

## Exercise 5: Type-driven expression building

- Create variables for name (str), age (int), height (float), active (bool).
- Print all four variables in one line using commas inside `print()`.
- Create `first_name` and `last_name`, then print the full name using `+` and a space.
- Add `age` and `height`, print the result and its type.
- Print the type of every variable at the end.

```python
# 1. Create variables
name = "Alex"
age = 28
height = 5.9
active = True

# 2. Print all four variables in one line using commas
print(name, age, height, active)

# 3. Print full name using '+' and space
first_name = "Alex"
last_name = "Morgan"
full_name = first_name + " " + last_name
print(full_name)

# 4. Add age and height, print result and type
age_height_sum = age + height
print("Sum:", age_height_sum)
print("Type of sum:", type(age_height_sum))

# 5. Print types of all variables
print("\n--- Types ---")
print("name:", type(name))
print("age:", type(age))
print("height:", type(height))
print("active:", type(active))
print("first_name:", type(first_name))
print("last_name:", type(last_name))
print("full_name:", type(full_name))
```
