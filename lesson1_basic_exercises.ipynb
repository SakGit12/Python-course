## Exercise 3: Multiple assignment and unpacking

1. Assign `x`, `y`, and `z` in one line with values `10`, `"Hi"`, and `3.5`.

````python
# 1. Assign x, y, z in one line with different values
x, y, z = 10, "Hi", 3.5

# 2. Assign the value 100 to a, b, and c in one line
a = b = c = 100

# 3. Unpack list into three variables
colors = ["red", "green", "blue"]
color1, color2, color3 = colors

# Print all variables to verify
print("x, y, z:", x, y, z)
print("a, b, c:", a, b, c)
print("Unpacked colors:", color1, color2, color3)
````

## Exercise 5: Variable reassignment + data types

1. Create a variable `item = "book"` and print it.
2. Reassign `item` to `50` and print it again.
3. Reassign `item` to `True` and print it again.
4. Create one value for each type below and print `type(...)` for each:
   - `str`
   - `int`
   - `float`
   - `list`
   - `dict`
   - `bool`
   - `NoneType`

````python
# 1. Create a variable item = "book" and print it
item = "book"
print("Initial item:", item)

# 2. Reassign item to 50 and print it again
item = 50
print("Reassigned to int:", item)

# 3. Reassign item to True and print it again
item = True
print("Reassigned to bool:", item)

# 4. Create one value for each type and print type(...) for each
val_str = "Python"
val_int = 42
val_float = 3.14
val_list = [1, 2, 3]
val_dict = {"name": "Alice", "age": 20}
val_bool = False
val_none = None

print("\n--- Data Types ---")
print("str:", type(val_str))
print("int:", type(val_int))
print("float:", type(val_float))
print("list:", type(val_list))
print("dict:", type(val_dict))
print("bool:", type(val_bool))
print("NoneType:", type(val_none))
````
