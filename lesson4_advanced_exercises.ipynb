## Exercise 1: Nested list processing

- Create a 2D list (3 rows x 4 cols) of numbers.
- Print row sums and grand total.
- Print the maximum value from each row.

```python
matrix = [
    [4, 8, 3, 9],
    [7, 2, 6, 1],
    [5, 10, 11, 3]
]
row_sums = [sum(row) for row in matrix]
print("Row sums:", row_sums)
print("Grand total:", sum(row_sums))
print("Max in each row:", [max(row) for row in matrix])
```

## Exercise 2: Comprehension with condition

- From numbers 1..30, build a list of numbers divisible by 3 but not by 2.
- Then build a second list of their squares using comprehension.
- Print both lists.

```python
numbers = list(range(1, 31))
filtered = [n for n in numbers if n % 3 == 0 and n % 2 != 0]
squares = [n ** 2 for n in filtered]
print("Filtered numbers:", filtered)
print("Squares:", squares)
```

## Exercise 3: Custom sorting

- Create a list of lists: `[name, score]` for at least 5 students.
- Write a helper function that returns a sort key `[-score, name]`.
- Sort descending by score, then ascending by name for ties, using `sort(key=...)`.
- Print the sorted result.

```python
students = [
    ["Lina", 90],
    ["Sam", 84],
    ["Alice", 90],
    ["Milo", 76],
    ["Zoe", 84]
]
def sort_key(student):
    return (-student[1], student[0])
students.sort(key=sort_key)
print("Sorted students:", students)
```

## Exercise 4: Duplicate removal (order preserved)

- Given a list with duplicates, create a new list without duplicates while preserving original order.
- Use a loop and a membership check (`not in`) before appending.
- Print original and cleaned lists.

```python
values = [3, 1, 3, 2, 1, 4, 2, 5]
cleaned = []
for value in values:
    if value not in cleaned:
        cleaned.append(value)
print("Original:", values)
print("Cleaned:", cleaned)
```

## Exercise 5: Merge and flatten

- Create two lists of lists, merge them, and flatten into one list.
- Print merged nested list and flattened list.
- Compute and print average of flattened numeric values.

```python
list1 = [[1, 2], [3, 4]]
list2 = [[5, 6], [7]]
merged = list1 + list2
flat = [num for sublist in merged for num in sublist]
print("Merged:", merged)
print("Flattened:", flat)
print("Average:", sum(flat) / len(flat))
```
