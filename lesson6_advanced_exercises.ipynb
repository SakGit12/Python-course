## Exercise 1: Dictionary merge with conflict rule

- Create two dictionaries with overlapping keys.
- Merge them so that larger numeric value wins for overlapping keys.
- Print merged dictionary.

```python
dict1 = {"a": 10, "b": 5, "c": 7}
dict2 = {"b": 8, "c": 12, "d": 3}
merged = dict1.copy()
for key, value in dict2.items():
    if key in merged:
        merged[key] = max(merged[key], value)
    else:
        merged[key] = value
print("Merged dictionary:", merged)
```

## Exercise 2: Nested dictionary navigation

- Create a nested dictionary representing 3 students with scores in 3 subjects.
- Print each student's average score.
- Print student name with highest average.

```python
students = {
    "Alice": {"Math": 90, "Science": 85, "English": 88},
    "Bob": {"Math": 75, "Science": 80, "English": 82},
    "Cara": {"Math": 95, "Science": 92, "English": 91}
}
averages = {}
for name, scores in students.items():
    averages[name] = sum(scores.values()) / len(scores)
    print(f"{name}: {averages[name]:.2f}")
best_student = max(averages, key=averages.get)
print("Highest average:", best_student, averages[best_student])
```

## Exercise 3: Dictionary comprehension

- From list `words = ["apple", "banana", "kiwi", "grape"]`, build dictionary `{word: length}`.
- Build second dictionary with only words of length >= 5.
- Print both dictionaries.

```python
words = ["apple", "banana", "kiwi", "grape"]
word_lengths = {word: len(word) for word in words}
long_words = {word: len(word) for word in words if len(word) >= 5}
print("All lengths:", word_lengths)
print("Long words:", long_words)
```

## Exercise 4: Safe access and defaults

- Given dictionary with optional keys, use `.get()` with defaults.
- Print fallback values for missing keys.
- Avoid KeyError in your solution.

```python
profile = {"name": "Nora", "city": "Phnom Penh"}
print("Name:", profile.get("name", "Unknown"))
print("Age:", profile.get("age", "Not provided"))
print("City:", profile.get("city", "Unknown"))
```

## Exercise 5: Invert mapping

- Given dictionary mapping student->grade, create grade->list_of_students.
- Sort each student list alphabetically.
- Print inverted mapping.

```python
student_grades = {"Alice": "A", "Bob": "B", "Cara": "A", "Dara": "C"}
grade_to_students = {}
for student, grade in student_grades.items():
    grade_to_students.setdefault(grade, []).append(student)
for grade in grade_to_students:
    grade_to_students[grade].sort()
print("Inverted mapping:", grade_to_students)
```
