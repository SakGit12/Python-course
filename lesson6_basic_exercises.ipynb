## Exercise 1: Dictionary basics

1. Create:

```python
student = {
    "name": "Anna",
    "age": 21,
    "course": "Python"
}
```

2. Print the whole dictionary.
3. Print `name` and `course` values.
4. Print the number of key-value pairs.

````python
student = {"name": "Anna", "age": 21, "course": "Python"}
print(student)
print(student["name"])
print(student["course"])
print("Number of items:", len(student))
````

## Exercise 2: Add, update, remove

Using `student`:

1. Add key `"city"` with value `"PhnomPenh"`.
2. Update `"age"` to `22`.
3. Remove `"course"` using `pop()`.
4. Print dictionary after each change.

````python
student = {"name": "Anna", "age": 21, "course": "Python"}
student["city"] = "PhnomPenh"
print("After add:", student)
student["age"] = 22
print("After update:", student)
student.pop("course")
print("After remove:", student)
````

## Exercise 3: Loop through dictionary

1. Create `product = {"name": "Laptop", "price": 1200, "stock": 5}`.
2. Loop through keys and print each key.
3. Loop through values and print each value.
4. Loop through items and print `key: value`.

````python
product = {"name": "Laptop", "price": 1200, "stock": 5}
for key in product:
    print(key)
for value in product.values():
    print(value)
for key, value in product.items():
    print(f"{key}: {value}")
````

## Exercise 4: Copy dictionary

1. Create `original = {"a": 1, "b": 2}`.
2. Make a copy using `copy()` and store in `copied`.
3. Change `original["a"]` to `99`.
4. Print both dictionaries to show they are separate.

````python
original = {"a": 1, "b": 2}
copied = original.copy()
original["a"] = 99
print("Original:", original)
print("Copied:", copied)
````

## Exercise 5: Nested dictionaries

1. Create a nested dictionary `school` with two students (`student1`, `student2`), each with `name` and `grade`.
2. Print only `student2` name.
3. Print all keys using `keys()`.
4. Print all values using `values()`.
5. Print all key-value pairs using `items()`.

````python
school = {
    "student1": {"name": "Sok", "grade": "A"},
    "student2": {"name": "Mali", "grade": "B"}
}
print("student2 name:", school["student2"]["name"])
print("Keys:", school.keys())
print("Values:", school.values())
print("Items:", school.items())
````
